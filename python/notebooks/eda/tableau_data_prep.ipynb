# Purpose of This Notebook

This notebook prepares curated, analysis-ready datasets for visualization and reporting in Tableau.

After completing exploratory data analysis, statistical testing, and predictive modeling, this notebook serves as the final data layer that translates our findings into clear, stakeholder-facing visuals.

## Why This Matters
- Our Critical Research Question asks: Do neighborhoods with higher shelter-related food insecurity prioritization have sufficient emergency food assistance supply?
- To answer this effectively for policymakers and stakeholders, we need structured visual outputs that show:
    - 📍 Neighborhood-level need (prioritization scores, food insecurity rates)
    - 🥫 Supply of EFAP sites (total sites, pantry vs kitchen, weekend access)
    - 🏠 Shelter demand context (families with children in shelters)
    - ⚖️ Mismatch between high need and low supply
- This notebook generates the SQL queries used to extract and join these components from our star schema database


## What This Notebook Does
**Connects to the food_access.db database and joins fact and dimension tables:**
- fact_food_site_coverage
- fact_neighborhood_prioritization
- fact_agg_shelter_cdta_year
- dim_map
- dim_date

**Produces Tableau-ready datasets for:**
- High-need vs supply comparisons
- Underserved neighborhood rankings
- Shelter demand vs EFAP supply analysis
- Borough-level summaries


## workflow
- EDA → Statistical Analysis → Modeling → Dashboard Data Layer (This Notebook) → Tableau Visualization → Final Presentation

### Visual 1) high need neighborhoods vs supply
- Chart idea: Bar chart or scatter: prioritization (need) vs total sites (supply)  
- Tableau use:
    - x-axis: weighted_score or food_insecure_percentage
    - y-axis: total_sites
    - color: boro_name or supply_gap
    - filter: year

```python
query_1 = """
SELECT
  m.boro_name,
  f.nta_id,
  m.nta_name,
  d.year,
  p.weighted_score,
  p.food_insecure_percentage,
  p.supply_gap,
  f.total_sites,
  f.pantry_site_count,
  f.kitchen_site_count,
  f.weekend_site_count
FROM fact_food_site_coverage f
JOIN dim_date d ON d.date_id = f.date_id
JOIN dim_map m ON m.nta_id = f.nta_id
JOIN fact_neighborhood_prioritization p ON p.nta_id = f.nta_id
WHERE d.year IN (2023, 2024);
"""
df_need_vs_supply = pd.read_sql_query(query_1, conn)

df_need_vs_supply.head()
```

### Visual 2: The mismatch list (top underserved NTAs)
- Chart idea: ranked table or bar: “highest need, lowest supply”

```python
query_2 = """
SELECT
  d.year,
  m.boro_name,
  f.nta_id,
  m.nta_name,
  p.weighted_score,
  p.food_insecure_percentage,
  p.supply_gap,
  f.total_sites,
  f.kitchen_site_count,
  f.weekend_site_count
FROM fact_food_site_coverage f
JOIN dim_date d ON d.date_id = f.date_id
JOIN dim_map m ON m.nta_id = f.nta_id
JOIN fact_neighborhood_prioritization p ON p.nta_id = f.nta_id
WHERE d.year IN (2023, 2024)
ORDER BY p.weighted_score DESC, f.total_sites ASC
LIMIT 25;
"""

df_mismatch = pd.read_sql_query(query_2, conn)

df_mismatch.head()
```

### Visual 3: Shelter demand context added (CDTA families)
- Chart idea: Scatterplot: shelter families vs total sites (supply)
- Tableau use:
    - x-axis: family_with_children_count
    - y-axis: total_sites
    - color: weighted_score
    - filter: borough/yeair


```python
query_3 = """ 
WITH shelter AS (
  SELECT
    s.cdta_id, d.year, s.family_with_children_count
  FROM fact_agg_shelter_cdta_year s
  JOIN dim_date d ON d.date_id = s.date_id
  WHERE d.year IN (2023, 2024)
)
SELECT
  sh.year,
  m.cdta_id,
  m.cdta_name,
  m.boro_name,
  m.nta_id,
  m.nta_name,
  sh.family_with_children_count,
  f.total_sites,
  p.weighted_score
FROM shelter sh
JOIN dim_map m ON m.cdta_id = sh.cdta_id
JOIN fact_food_site_coverage f ON f.nta_id = m.nta_id
JOIN dim_date d2 ON d2.date_id = f.date_id AND d2.year = sh.year
JOIN fact_neighborhood_prioritization p ON p.nta_id = m.nta_id;

"""

df_cdta_demand_supply = pd.read_sql_query(query_3, conn)

df_cdta_demand_supply.head()
```

### Visual 4: Access type breakdown (kitchen vs pantry vs weekend)
- Chart idea: Stacked bar by borough or year
- Great for showing operational capacity differences by borough!!

```python
query_4 = """ 
SELECT
  d.year,
  m.boro_name,
  SUM(f.total_sites) AS total_sites,
  SUM(f.pantry_site_count) AS pantry_sites,
  SUM(f.kitchen_site_count) AS kitchen_sites,
  SUM(f.weekend_site_count) AS weekend_sites
FROM fact_food_site_coverage f
JOIN dim_date d ON d.date_id = f.date_id
JOIN dim_map m ON m.nta_id = f.nta_id
WHERE d.year IN (2023, 2024)
GROUP BY d.year, m.boro_name
ORDER BY d.year, m.boro_name;
"""

df_access_type = pd.read_sql_query(query_4, conn)
df_access_type.head()
```

### Visual 1: Where is need highest?

- **Purpose**: This query creates an NTA-level dataset for the Tableau map that highlights where food insecurity need is highest, while showing shelter family pressure (CDTA context) in the tooltip for 2024.
    - Grain: One row per NTA (nta_id).
    - Filters: 
        - np.is_high_priority = 1 (restricts to high-need NTAs)
        - d.year = 2024 (uses 2024 shelter family counts)
    - Joins:
        - `fact_neighborhood_prioritization` (np) -> provides need/prioritization measures at the NTA level.
        - `dim_map` (m) joined on nta_id -> adds NTA metadata (name, borough) and geometry (the_geom_wkt) needed for mapping.
        - `fact_agg_shelter_cdta_year` (agc) joined via cdta_id -> brings in shelter family counts at the community district level.
        - `dim_date` (d) joined via `date_id` ->  enables year filtering (2024).
    - Fields returned
        - Map keys / geography
        - `nta_id, nta_name, boro_name, cdta_id, the_geom_wkt`
    - Need indicators (tooltip + color)
        - `weighted_score` (used for __color__)
        - `food_insecure_percentage`
        - `supply_gap`
        - `vulnerable_pop_percentage`
        - Shelter pressure indicator (tooltip)
- family_with_children_count_2024 = SUM(agc.family_with_children_count)
    - Why SUM is used: Shelter counts are stored at the CDTA level and joined to NTAs through dim_map. Using SUM() with GROUP BY prevents duplicate rows and ensures a single NTA record even if multiple shelter rows exist for a CDTA within the filtered year.

- **Intended Tableau Use**
- Map layer uses `the_geom_wkt`
- Color by `weighted_score`
- Tooltip includes shelter family count + additional need drivers to explain “pressure”

```python
query_1 = """
WITH shelter_by_cdta AS (
  SELECT
    agc.cdta_id,
    SUM(agc.family_with_children_count) AS family_with_children_count_2024
  FROM fact_agg_shelter_cdta_year agc
  JOIN dim_date d
    ON d.date_id = agc.date_id
  WHERE d.year = 2024
  GROUP BY agc.cdta_id
),
cdta_ranked AS (
  SELECT
    cdta_id,
    family_with_children_count_2024,
    CUME_DIST() OVER (ORDER BY family_with_children_count_2024) AS cdta_cume
  FROM shelter_by_cdta
),
cdta_flags AS (
  SELECT
    cdta_id,
    family_with_children_count_2024,
    CASE WHEN cdta_cume >= 0.75 THEN 1 ELSE 0 END AS is_high_shelter
  FROM cdta_ranked
)

SELECT
  np.nta_id,
  m.nta_name,
  m.boro_name,
  m.cdta_id,
  m.the_geom_wkt,

  -- Need / prioritization metrics
  np.weighted_score,
  np.food_insecure_percentage,
  np.supply_gap,
  np.vulnerable_pop_percentage,
  np.is_high_priority,

  -- Shelter context (2024) + flag
  COALESCE(cf.family_with_children_count_2024, 0) AS family_with_children_count_2024,
  COALESCE(cf.is_high_shelter, 0) AS is_high_shelter,

  -- Combined “pressure zone”
  CASE WHEN np.is_high_priority = 1 AND COALESCE(cf.is_high_shelter, 0) = 1 THEN 1 ELSE 0 END AS is_pressure_zone

FROM fact_neighborhood_prioritization np
JOIN dim_map m
  ON m.nta_id = np.nta_id
LEFT JOIN cdta_flags cf
  ON cf.cdta_id = m.cdta_id;
"""

df_pressure = pd.read_sql_query(query_1, conn)
df_pressure.head()
df_pressure.to_csv("../../../data/tableau_use/visual_1_need_pressure_2024.csv", index=False)
```

```python
import geopandas as gpd
from shapely import wkt
df_pressure = pd.read_sql_query(query_1, conn)

# Convert WKT string to geometry objects
df_pressure["geometry"] = df_pressure["the_geom_wkt"].apply(wkt.loads)

# Convert to GeoDataFrame
gdf_pressure = gpd.GeoDataFrame(df_pressure, geometry="geometry")

# Drop original WKT column (optional but cleaner)
gdf_pressure = gdf_pressure.drop(columns=["the_geom_wkt"])

# Set coordinate reference system (NYC NTAs are usually WGS84)
gdf_pressure = gdf_pressure.set_crs(epsg=4326)

# Export as GeoJSON
gdf_pressure.to_file("../../../data/tableau_use/visual_1_need_pressure_2024.geojson", driver="GeoJSON")
```

After running the SQL query, the map shapes were stored as text in something called WKT, which is basically just coordinates written out as a long string. Tableau cannot read that text as an actual map shape. So I converted that text into real geometry objects using Shapely, which turns those coordinate strings into actual polygons. Then I told Python what coordinate system the shapes use so they align correctly on a map. After that, I removed the original text column since it was no longer needed and exported everything as a GeoJSON file. GeoJSON is a format Tableau understands as spatial data, so this allows the NTAs to render properly as map polygons while keeping all the neighborhood metrics attached.
